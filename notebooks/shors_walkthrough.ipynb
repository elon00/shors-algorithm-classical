# Shor's Algorithm — Complete Walkthrough

This notebook walks through every classical subroutine in Shor's algorithm,
from GCD to the full factoring algorithm. Run each cell sequentially.

```python
import sys
sys.path.insert(0, '..')

from shors_classical.gcd import gcd, extended_gcd
from shors_classical.modular_arithmetic import modular_exponentiate, modular_inverse
from shors_classical.continued_fractions import (
    continued_fraction_expansion_rational,
    convergent_from_cf,
    find_period_from_continued_fraction,
)
from shors_classical.period_finding import find_period_brute_force
from shors_classical.shor_classical import shors_algorithm_classical
```

## 1. GCD — The Building Block

```python
print(f"gcd(48, 18) = {gcd(48, 18)}")
print(f"gcd(17, 5) = {gcd(17, 5)}")
g, x, y = extended_gcd(48, 18)
print(f"Extended: 48*{x} + 18*{y} = {g}")
```

## 2. Modular Exponentiation

```python
print(f"2^10 mod 1000 = {modular_exponentiate(2, 10, 1000)}")
print(f"3^5 mod 7 = {modular_exponentiate(3, 5, 7)}")
print(f"7^4 mod 15 = {modular_exponentiate(7, 4, 15)}  (period r=4)")
```

## 3. Continued Fractions — Extract the Period

```python
Q, N, r = 256, 15, 4
for j in range(1, r):
    m = round(j * Q / r)
    cand = find_period_from_continued_fraction(m, Q, N)
    print(f"j={j}: m/Q = {m}/{Q}, candidates r = {cand}")
```

## 4. Brute-Force Period Finding

```python
for a in [2, 7]:
    r = find_period_brute_force(a, 15)
    print(f"Period of {a} mod 15: r = {r}")
    seq = [modular_exponentiate(a, x, 15) for x in range(2*r)]
    print(f"  Sequence: {seq}")
```

## 5. Full Shor's Algorithm

```python
for N in [15, 21, 35]:
    result = shors_algorithm_classical(N, verbose=False)
    print(f"{N} = {result} x {N // result}")
```

## 6. Simulated Quantum Period Finding

```python
print("Using continued fractions (quantum simulation):")
result = shors_algorithm_classical(15, use_quantum_simulation=True, verbose=True)
```
